```python
# Step 2: Input variables (Conditions)
base_price = 20.0  # Chips packet-oda normal price (Rs.20)
is_raining = True  # Mazhai peyiratha? (True / False)
stock_level = "Low" # Godown-la stock epdi irukku ("Low", "Medium", "High")
competitor_price = 22.0 # Competitor antha packet-ah evloku vikkuraanga
```

```python
# Step 3: Dynamic Pricing Logic Function
def calculate_dynamic_price(base_price, is_raining, stock_level, competitor_price):
    final_price = base_price

    # Condition 1: Mazhai peyiruntha delivery demand adhigamum, price 20% increase aagum
    if is_raining:
        final_price = final_price * 1.20 # Rs. 24 aagum

    # Condition 2: Stock romba kammiya irundha price innum 10% increase aagum
    if stock_level == "Low":
        final_price = final_price * 1.10

    # Condition 3: Competitor price romba high-a irundha, namma price-ah competitor price-ku aathukalam (Max limit)
    if final_price > competitor_price:
        final_price = competitor_price

    return round(final_price, 2)
```

```python
# Step 4: Run the Engine
new_price = calculate_dynamic_price(base_price, is_raining, stock_level, competitor_price)

print(f"Base Price: Rs. {base_price}")
print(f"Calculated Dynamic Price: Rs. {new_price}")
```

```python
import pandas as pd

# 1. Oru sample dataset create panrathu (Multiple products)
data = {
    'Product_Name': ['Chips', 'Biscuit', 'Chocolate', 'Cold Drink'],
    'Base_Price': [20.0, 10.0, 30.0, 40.0],
    'Is_Raining': [True, False, True, False],
    'Stock_Level': ['Low', 'High', 'Low', 'Medium'],
    'Competitor_Price': [22.0, 12.0, 35.0, 45.0]
}

df = pd.DataFrame(data)

# 2. Dynamic Price calculate panra function-ah DataFrame-ku apply panrathu
def apply_dynamic_pricing(row):
    price = row['Base_Price']
    if row['Is_Raining']:
        price *= 1.20
    if row['Stock_Level'] == 'Low':
        price *= 1.10
    if price > row['Competitor_Price']:
        price = row['Competitor_Price']
    return round(price, 2)

# 3. Pudhu column-ah add panrathu
df['Dynamic_Price'] = df.apply(apply_dynamic_pricing, axis=1)

# Result-ah table form-la pakrathu
print(df)

# Idha oru CSV file-ah download panna:
df.to_csv('dynamic_pricing_output.csv', index=False)
```
